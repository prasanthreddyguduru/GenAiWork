```python
# write your solution here
nums=[1,2,3,4,5,6]
nums=[num*num for num in nums if num%2==0]
print (nums)
```

- Q4.1: Write a function calculate_area(length, width) that returns the area of a rectangle. If only one argument is provided, assume it is a square (width = length).

    - Expected Output for (5, 4): 20

    - Expected Output for (5): 25

```python
# write your solution here
def calculate_area(length,width=None) :
    if width==None :
        return length * length
    else :
        return length*width;
print (calculate_area(4))
```
